# Bibliotheken importeren

Importeer Matplotlib voor visualisatie en NumPy voor het genereren van numerieke waarden.

```python
import matplotlib.pyplot as plt
import numpy as np
```

# Energieverbruiksfunctie definiëren

Definieer de functie y(x) = 2x + 1, die het energieverbruik berekent op basis van de invoerwaarde x.

```python
def y(x):
    return 2 * x + 1
```

# Datapunten genereren

Gebruik np.linspace() om 500 x-waarden tussen 0 en 10 te genereren en bereken de bijbehorende y-waarden met de energieverbruiksfunctie.

```python
x_values = np.linspace(0, 10, 500)
y_values = y(x_values)
```

# Energieverbruik visualiseren

Maak een Matplotlib-lijngrafiek met een titel, aslabels, begrensde assen, rasterlijnen en een legenda.

```python
plt.figure(figsize=(10, 6))
plt.plot(x_values, y_values, label="y = 2*x + 1")
plt.title("Energieverbruik van een hardloper")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim([0, 5])
plt.ylim([0, 10])
plt.grid(True)
plt.legend(loc="upper left")
plt.show()
```
