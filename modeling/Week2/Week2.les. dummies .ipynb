```python
import pandas as pd

insurance = pd.read_csv('insurance.csv')
insurance= pd.get_dummies(insurance, dtype=int)
insurance.head()
```

```python
import statsmodels.api as sm

y = insurance['charges'] #define the dependent variable
x = insurance [['sex_male', 'smoker_yes' ]]
x = sm.add_constant(x)

model = sm.OLS(y,x).fit()
print(model.summary())
```
