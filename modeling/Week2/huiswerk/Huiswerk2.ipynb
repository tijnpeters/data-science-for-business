```python
import pandas as pd
import matplotlib.pyplot as plt
wine = pd.read_csv('winequality-red.csv')
data =wine[['fixed_acidity', 'volatile_acidity', 'citric_acid', 'residual_sugar', 'chlorides', 'free_sulfur_dioxide', 'total_sulfur_dioxide', 'density', 'pH', 'sulphates', 'alcohol', 'quality']]
data.head()
```

```python
y = wine['quality']
x = wine[['fixed_acidity', 'volatile_acidity', 'citric_acid', 'residual_sugar', 'chlorides', 'free_sulfur_dioxide', 'total_sulfur_dioxide', 'density', 'pH', 'sulphates', 'alcohol']]
x=sm.add_constant(x)
model=sm.OLS(y,x).fit()
print(model.summary())
```
