```python
import pandas as pd
import matplotlib.pyplot as plt
happiness = pd.read_csv('World Happiness 2019.csv')
data = happiness[['Happiness_Score', 'GDP_per_capita']]
data.head
```

```python
import statsmodels.api as sm
```

```python
y = happiness['Happiness_Score']
x = happiness[['GDP_per_capita','Social_support','Freedom_to_make_life_choices','Generosity','Perceptions_of_corruption']]
x=sm.add_constant(x)
model=sm.OLS(y,x).fit()
print(model.summary())
```

**1A: There is a moderatative positive correlation between the variables**

**1B,C: The multiple regression model explains 76.4% of the variation in Happiness Score (R² = 0.764). GDP per capita, social support and freedom to make life choices are statistically significant predictors (p < 0.05), while generosity and perceptions of corruption are not significant at the 5% level.**
