```python
import pandas as pd

bank = pd.read_csv('bank.csv')
bank= pd.get_dummies(bank, dtype=int)
bank.head()
```

```python
import statsmodels.api as sm

y = bank['charges'] #define the dependent variable
x = bank [['sex_male', 'smoker_yes' ]]
x = sm.add_constant(x)

model = sm.OLS(y,x).fit()
print(model.summary())
```
