```python
import pandas as pd
import matplotlib.pyplot as plt
happiness = pd.read_csv('World Happiness 2019.csv')
data = happiness[['Happiness_Score', 'GDP_per_capita']]
data.head
```

```python
df = pd.read_csv('World Happiness 2019.csv')
df.head()
```

```python
plt.scatter(df['Happiness_Score'],df['GDP_per_capita'])
plt.xlabel('Happiness Score')
plt.ylabel('GDP per Capita')
plt.title('Relationship between Happiness and GDP')
plt.show()
```

What I is that there is a positive relationship between de variables but it's very spread out

```python
correlation = df['Happiness_Score'].corr(df['GDP_per_capita'])
 
print(correlation)
```

The correlation is 0.79 what means that there is a moderate positive correlation between the variables

```python
import pandas as pd
videogames = pd.read_excel('videogames.xlsx')
print(videogames.head())
```

```python
plt.scatter(videogames['Sales'],videogames['Used_Price'])
plt.xlabel('Sales')
plt.ylabel('Used Price')
plt.title('Relationship between Sales and Used Price')
plt.show()
```

```python
import statsmodels.api as sm

y = videogames['Sales'] #define the dependent variable
X = videogames[['Used_Price','Review_Score','Release_Year','Length_All_PlayStyles_Polled']] #define the independent variable
X = sm.add_constant(X) #add an intercept (DON'T FORGET!)

model1 = sm.OLS(y,X) #create the model by putting in the variables (has to be in this order!)
results = model1.fit() #produce results by fitting the model to the data
print(results.summary()) #print a summary of the model results
```
