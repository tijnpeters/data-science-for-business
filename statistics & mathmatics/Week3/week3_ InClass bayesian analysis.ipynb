***You’re the owner of a web shop and last year the conversion rate was 3%. The
conversion rate is the probability that a visitor makes a purchase. You made some
changes to the website and tested it on 1000 customers. In this sample 38 customers
ended up making a purchase.***


***a. What is the probability that at least 38 customers made a purchase, assuming the
conversation rate has stayed the same?***

k= 37 want je wil vanaf 37
n=1000 want 1000 customers
p= 0.3 want 3%

```python
from scipy.stats import binom
1 - binom.cdf(k=37, n=1000, p=0.3)
```
